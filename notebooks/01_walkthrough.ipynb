# `unsupervised-patterns`: Feature Extraction & Clustering

Reference implementations of unsupervised feature extraction: PCA (SVD & Eigendecomposition), K-Means++, Gaussian Mixture Models (EM), DBSCAN, and cluster validation metrics.

```python
import numpy as np
from sklearn.datasets import load_iris, load_wine, make_blobs, make_moons
from unsupervised_patterns.decomposition import PCA
from unsupervised_patterns.clustering import KMeans, GaussianMixture, DBSCAN
from unsupervised_patterns.metrics import silhouette_score, davies_bouldin_score, find_elbow_point
```

## 1. PCA: Dimensionality Reduction via SVD and Eigendecomposition

```python
wine = load_wine()
X_wine = wine.data

pca = PCA(n_components=3, method='svd').fit(X_wine)
print(f"Explained Variance Ratio: {pca.explained_variance_ratio_}")
print(f"Cumulative Variance: {np.cumsum(pca.explained_variance_ratio_)}")
```

## 2. K-Means++ Clustering & Metric Validation

```python
X_blobs, _ = make_blobs(n_samples=300, centers=4, cluster_std=0.8, random_state=42)
km = KMeans(n_clusters=4, init='k-means++', random_state=42).fit(X_blobs)

sil = silhouette_score(X_blobs, km.labels_)
db = davies_bouldin_score(X_blobs, km.labels_)
print(f"KMeans Inertia: {km.inertia_:.2f}")
print(f"Silhouette Score: {sil:.4f}")
print(f"Davies-Bouldin Index: {db:.4f}")
```

## 3. DBSCAN on Arbitrary Geometries (Moons Manifold)

```python
X_moons, _ = make_moons(n_samples=250, noise=0.08, random_state=42)
dbscan = DBSCAN(eps=0.2, min_samples=5).fit(X_moons)
n_clusters = len(set(dbscan.labels_) - {-1})
print(f"DBSCAN detected {n_clusters} clusters with {len(dbscan.core_sample_indices_)} core samples.")
```
